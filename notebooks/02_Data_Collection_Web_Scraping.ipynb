# 2. Data Collection — Web Scraping

The IBM capstone teaches collecting historical Falcon 9 launch records from Wikipedia using BeautifulSoup. This reconstruction uses a local HTML snapshot generated from the same course-era records so the notebook can be executed without relying on a live website.

```python
from bs4 import BeautifulSoup
from pathlib import Path
import pandas as pd
html=Path('../data/falcon9_wikipedia_snapshot.html').read_text()
soup=BeautifulSoup(html,'html.parser')
table=soup.find('table')
rows=[]
headers=[th.get_text(strip=True) for th in table.find_all('th')]
for tr in table.find_all('tr')[1:]:
    cells=[td.get_text(strip=True) for td in tr.find_all('td')]
    if cells: rows.append(cells)
scraped=pd.DataFrame(rows, columns=headers)
scraped.head()
```

```python
scraped['FlightNumber']=pd.to_numeric(scraped['FlightNumber'])
scraped['PayloadMass']=pd.to_numeric(scraped['PayloadMass'])
scraped['Class']=pd.to_numeric(scraped['Class'])
print('Scraped rows:',len(scraped))
scraped.tail()
```

```python
scraped.to_csv('../data/spacex_web_scraped.csv',index=False)
print('Saved spacex_web_scraped.csv')
```

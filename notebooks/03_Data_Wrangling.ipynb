# 3. Data Wrangling

Clean the Falcon 9 dataset, create the binary landing target, handle missing payload mass, and inspect data quality.

```python
import pandas as pd
df=pd.read_csv('../data/dataset_part_2.csv')
df.isna().sum().sort_values(ascending=False).head(10)
```

```python
# Payload mass: replace missing values with the observed mean, following the course workflow.
df['PayloadMass']=df['PayloadMass'].fillna(df['PayloadMass'].mean())
# The target Class is already encoded as 1=successful landing and 0=unsuccessful/no landing.
df['Date']=pd.to_datetime(df['Date'])
print(df[['PayloadMass','Class']].describe())
print('Target counts:')
print(df['Class'].value_counts())
```

```python
# Compact analysis-ready table
analysis_cols=['FlightNumber','Date','PayloadMass','Orbit','LaunchSite','Flights','GridFins','Reused','Legs','Block','ReusedCount','Serial','Longitude','Latitude','Class']
clean=df[analysis_cols].copy()
clean.to_csv('../data/clean_spacex.csv',index=False)
clean.head(10)
```

```python
print('Clean shape:',clean.shape)
print('Missing values after cleaning:',int(clean.isna().sum().sum()))
```

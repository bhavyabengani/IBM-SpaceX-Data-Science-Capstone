# 6. Interactive Visual Analytics — Folium

Map launch sites and inspect the relationship between geography and landing outcomes.

```python
import pandas as pd
import folium
from folium.plugins import MarkerCluster
df=pd.read_csv('../data/clean_spacex.csv')
center=[df['Latitude'].mean(),df['Longitude'].mean()]
m=folium.Map(location=center,zoom_start=4,tiles='OpenStreetMap')
cluster=MarkerCluster().add_to(m)
for _,r in df.iterrows():
    popup=f"Flight {int(r.FlightNumber)} | {r.LaunchSite} | {'Success' if r.Class else 'Failure'}"
    folium.CircleMarker([r.Latitude,r.Longitude],radius=5,tooltip=popup,popup=popup).add_to(cluster)
m.save('../plots/spacex_launch_sites_map.html')
m
```

```python
site=df.groupby('LaunchSite').agg(launches=('FlightNumber','count'),success_rate=('Class','mean'),latitude=('Latitude','first'),longitude=('Longitude','first')).reset_index()
site['success_rate_pct']=(site['success_rate']*100).round(1)
site.sort_values('success_rate_pct',ascending=False)
```

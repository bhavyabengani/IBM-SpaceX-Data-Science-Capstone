# 7. Plotly Dash Dashboard

The Dash application provides filters for launch site and displays total launches and successful landings. The executable app is in `app.py`.

```python
import pandas as pd
import plotly.express as px
df=pd.read_csv('../data/spacex_launch_dash.csv')
summary=df.groupby('LaunchSite')['Class'].agg(['count','sum']).reset_index().rename(columns={'count':'Launches','sum':'SuccessfulLandings'})
summary
```

```python
fig=px.pie(summary,names='LaunchSite',values='SuccessfulLandings',title='Successful landings by launch site')
fig.show()
```

```python
fig2=px.scatter(df,x='FlightNumber',y='Class',color='LaunchSite',title='Launch outcome by flight number')
fig2.show()
```

```python
print('Dashboard controls: Launch-site dropdown and outcome view.')
print('The app.py file can be run locally with Dash installed.')
```

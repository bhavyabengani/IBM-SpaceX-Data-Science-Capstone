# 1. Data Collection — SpaceX REST API

Independent reconstruction of the IBM Applied Data Science Capstone data-collection workflow. The original course uses the SpaceX REST API; this notebook also loads the archived course-era extract so the notebook remains reproducible offline.

```python
import pandas as pd
from pathlib import Path
DATA=Path('../data/dataset_part_1.csv')
api_url='https://api.spacexdata.com/v4/launches/past'
print('API endpoint:', api_url)
raw=pd.read_csv(DATA)
raw.head()
```

```python
print('Rows:', len(raw))
print('Columns:', list(raw.columns))
raw.info()
```

```python
# In the live course workflow, requests.get(api_url).json() is normalized into a dataframe.
# Here the archived extract is used to make the notebook deterministic.
data=raw.copy()
data['Date']=pd.to_datetime(data['Date'])
data.head(10)
```

```python
print('Date range:', data['Date'].min().date(), 'to', data['Date'].max().date())
print('Launch sites:', data['LaunchSite'].nunique())
print('Unique booster serials:', data['Serial'].nunique())
```

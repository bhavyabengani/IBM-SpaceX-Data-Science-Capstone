# 4. Exploratory Data Analysis with SQL

SQLite queries answer the capstone questions about launch sites, payloads, success rates, rankings, and time trends.

```python
import sqlite3, pandas as pd
df=pd.read_csv('../data/clean_spacex.csv')
con=sqlite3.connect(':memory:')
df.to_sql('spacex',con,index=False,if_exists='replace')
```

```python
q1='SELECT LaunchSite, COUNT(*) AS launches FROM spacex GROUP BY LaunchSite ORDER BY launches DESC'
pd.read_sql_query(q1,con)
```

```python
q2='SELECT LaunchSite, ROUND(AVG(PayloadMass),1) AS avg_payload_kg FROM spacex GROUP BY LaunchSite ORDER BY avg_payload_kg DESC'
pd.read_sql_query(q2,con)
```

```python
q3='SELECT LaunchSite, ROUND(100.0*AVG(Class),1) AS success_rate_pct FROM spacex GROUP BY LaunchSite ORDER BY success_rate_pct DESC'
pd.read_sql_query(q3,con)
```

```python
q4="SELECT strftime('%Y',Date) AS year, COUNT(*) AS launches, ROUND(100.0*AVG(Class),1) AS success_rate_pct FROM spacex GROUP BY year ORDER BY year"
pd.read_sql_query(q4,con)
```

```python
q5='SELECT LaunchSite, MAX(PayloadMass) AS max_payload_kg FROM spacex GROUP BY LaunchSite ORDER BY max_payload_kg DESC'
pd.read_sql_query(q5,con)
```
